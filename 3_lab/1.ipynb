## Обєктно-орієнтоване програмування (ООП)
- все в Пайтоно є об'єктами, включаючи числа, рядки та функції.
- якщо нам не вистарчає функціоналу, ми можемо створити власні об'єкти та класи.
- Клас - це шаблон для створення об'єктів. Клас визначає атрибути та методи, які будуть у об'єктів цього класу.
- Обєкт - це екземпляр класу (щось що створене на основі, за принципом цього класу). Коли ми створюємо об'єкт, ми створюємо екземпляр класу.

```python
i = int(1)
print(type(i), i.__class__)
```

- демонструємо відмінність між вбудованими класами та користувацькими класами, створеними за допомогою ключового слова class.
- Приклад: студенти та їх оцінки
В класичному підході - маємо словник, де ключ - це ім'я студента, а значення - його оцінка.

```python
students_marks = {
    "Любомир Л": {"ООП": 90, "Бази даних": 80, "Математика": None, "Українська мова": None},
    "Василь": {"ООП": 30, "Бази даних": 40, "Математика": None, "Українська мова": None},
    "Іван": {"ООП": 40, "Бази даних": 50, "Математика": 60, "Українська мова": None},
    "Петро": {"ООП": 50, "Бази даних": 60, "Математика": 70, "Українська мова": 80},
    "Любомир П": {"ООП": 60, "Бази даних": 70, "Математика": None, "Українська мова": None}
}

for student, subjects in students_marks.items():
    print(f"Студент {student} отримав оцінки:")
    print(f"\tOOП: {subjects['ООП']}")
    print(f"\tБази даних: {subjects['Бази даних']}")
    print(f"\tМатематика: {subjects['Математика']}")
    print(f"\tУкраїнська мова: {subjects['Українська мова']}")
    print(f"Рейтинг студента {student}: {(subjects['ООП'] + subjects['Бази даних']) / len(subjects)}")
```

переписати класичний підхід під об'єктно-орієнтоване програмування, створивши клас Student, який буде мати атрибути name та grade, а також методи для отримання та встановлення оцінки.

```python
class Student:
    """Це пустий шаблон класу, який можна використовувати для створення об'єктів студентів."""
    pass
# створюємо об'єкти згідно шаблону класу Student
std_1 = Student()
# Заповнюємо дані об'єкта std_1
std_1.name = "Любомир Л"
std_1.grade = {"ООП": 90, "Бази даних": 80, "Математика": None, "Українська мова": None}
std_1.birthday = "01.01.2000"
print(type(std_1))
print(std_1.birthday)  # це атрибут об'єкта std_1, який ми створили та заповнили даними

std_2 = Student()
std_2.name = "Василь"
std_2.grade = {"ООП": 30, "Бази даних": 40, "Математика": None, "Українська мова": None}
std_2.age = 20
print(type(std_2))
#print(std_2.birthday)  # тут буде помилка, бо ми не створювали атрибут birthday для об'єкта std_2

for student in [std_1, std_2]:
    print(f"Студент {student.name} отримав оцінки:")
    print(f"\tOOП: {student.grade['ООП']}")
    print(f"\tБази даних: {student.grade['Бази даних']}")
    print(f"\tМатематика: {student.grade['Математика']}")
    print(f"\tУкраїнська мова: {student.grade['Українська мова']}")
    print(f"Рейтинг студента {student.name}: {(student.grade['ООП'] + student.grade['Бази даних']) / len(student.grade)}")
```

створюємо клас, який тепер буде реальним прототипом для створення об'єктів студентів. Клас Student має атрибути name та grade, а також методи get_grade() та set_grade() для отримання та встановлення оцінки студента.
- в класі є особлива функція __init__, яка викликається при створенні об'єкта класу. Вона приймає параметри name та grade, які використовуються для ініціалізації атрибутів об'єкта.
- функції з подвійним підкресленням на початку та в кінці назви є спеціальними методами класу або магічними методами. 

```python
class Student:
    """Це клас, який представляє студента."""
    def __init__(self, name: str, grade: dict, birthday: str = None, age: int = None): # в конструктор передаються аргументи
        self.name = name # атрибут, це поле де можна записати значення та отримати його
        self.grades = grade
        self.birthday = birthday # атрибут, який можна заповнити пізніше
        self.age = age # атрибут, який можна заповнити пізніше

std_1 = Student(name="Любомир Л", grade={"ООП": 90, "Бази даних": 80, "Математика": None, "Українська мова": None})
print(std_1.birthday)  # тут буде None, бо ми не передавали значення для атрибуту birthday
std_2 = Student(name="Василь", grade={"ООП": 30, "Бази даних": 40, "Математика": None, "Українська мова": None})
print(std_2.age)

for student in [std_1, std_2]:
    print(f"Студент {student.name} отримав оцінки:")
    print(f"\tOOП: {student.grades['ООП']}")
    print(f"\tБази даних: {student.grades['Бази даних']}")
    print(f"\tМатематика: {student.grades['Математика']}")
    print(f"\tУкраїнська мова: {student.grades['Українська мова']}")
    print(f"Рейтинг студента {student.name}: {(student.grades['ООП'] + student.grades['Бази даних']) / len(student.grades)}")
```

- є атрибути динамічні та наперед задані (визначені в конструкторі). Динамічні атрибути можна додавати після створення об'єкта, а наперед задані визначаються в конструкторі класу.

```python
class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Любомир Л")
s1.grade = 3 # динамічний атрибут, який можна додати після створення об'єкта
s2 = Student("Василь")
s2.marks = 4

for m in [s1, s2]:
    print(m.name)
    # обєкти не є уніфікованими і ми не знаємо який з атрибутві існує
    if hasattr(m, 'grade'):
        print(m.grade)
    if hasattr(m, 'marks'):
        print(m.marks)
```

Атрибути можна перезаписувати, або вони є змінними

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.oop_grades = None
        self.db_grades = None
        self.math_grades = None
        self.ukr_grades = None

s1 = Student("Любомир Л")
s1.oop_grades = 90
s1.db_grades = 80
s2 = Student("Василь")
s2.oop_grades = 30
s2.db_grades = 40

s1.oop_grades = 100 # перезаписуємо значення атрибуту oop_grades для об'єкта s1
for m in [s1, s2]:
    print(f"Студент {m.name} отримав оцінки:")
    print(f"\tOOП: {m.oop_grades}")
    print(f"\tБази даних: {m.db_grades}")
    print(f"\tМатематика: {m.math_grades}")
    print(f"\tУкраїнська мова: {m.ukr_grades}")
    print(f"Рейтинг студента {m.name}: {(m.oop_grades + m.db_grades) / 4}")
```

- для того щоб читати атрибути і не могти їх модифікувати існують властивості (property). Властивість дозволяє створювати методи, які виглядають як атрибути. Властивості дозволяють контролювати доступ до атрибутів та забезпечують інкапсуляцію даних.
- з атрибутами ми працюємо всередині програми а властивості видаємо на зовні, щоб користувачі класу могли отримувати доступ до даних без можливості їх змінювати.

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.oop_grades = None # можна змінювати
        self.db_grades = None

    @property
    def get_oop_grade(self): # властивості можна тільки читати, але не можна змінювати
        return self.oop_grades

    @property
    def get_db_grade(self):
        return self.db_grades

s1 = Student("Любомир Л")
s1.oop_grades = 90
print(f"Оцінка з ООП: {s1.get_oop_grade}")
#s1.get_oop_grade = 100 # тут буде помилка, бо ми не можемо змінювати властивість get_oop_grade
```
