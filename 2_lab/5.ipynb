## Розгалуження

```python
import random
condition = random.choice([True, False])

print("Підкиданя монети")
if condition:
    print("Монета випала гербом")
else:
    print("Монета випала решкою")
```

- розгалуженя можуть бути скороченми, коли є лише одна дія, яка виконується у випадку істинності умови. У такому випадку можна не використовувати ключове слово else.

```python
action = random.choice([True, False])
if action:
    print("Дія виконується")
print("Продовження виконання програми")
```

- така скорочена умова найчастіше викорстовується в фнкціях з поєднанням з оператором return. У такому випадку, якщо умова істинна, функція повертає значення і виходить з неї, а якщо умова хибна, то виконання функції продовжується далі.

```python
def classic(num): 
    if num > 0:
        return "Число додатнє"
    else:
        return "Число не додатнє"

def check_number(num):
    if num > 0:
        return "Число додатнє"
    return "Число не додатнє"

print(classic(5), check_number(0))
```

- ми може створити розгалуження використовуючи списки та вбудовані константи

```python
lf = ["Хибність", "Істина"]
print(lf[False], lf[True])

for i in range(5):
    c = random.choice([True, False])
    print(lf[c])
```

```python
l_obj = [min, max]
print(type(l_obj[False]))

print(l_obj[False]([5, 10, 6, 12, 7, 8]), l_obj[True]([5, 10, 6, 12, 7, 8]))
print(min([5, 10, 6, 12, 7, 8]), max([5, 10, 6, 12, 7, 8]))
```

- one-liners - однострічкові вирази, які використовуються для скорочення коду. Вони дозволяють виконувати умовні операції в одному рядку, що робить код більш компактним та зручним для читання.

```python
condition = random.choice([True, False])
a = ""
if condition:
    a = "Умова істинна"
else:
    a = "Умова хибна"

print(a)

a = "Умова істинна" if condition else "Умова хибна"
print(a)
b = 1 if condition else 0
print(b)
```
