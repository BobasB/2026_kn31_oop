## Функції
- функція - це блок коду який перевикористовується для виконання певної задачі. Функції дозволяють організовувати код, роблячи його більш зрозумілим та легким для підтримки.
- якщо код вживається двічі - значить його потрібно винести в функцію, щоб не дублювати код.
- приклад: визначення категорії оцінки за балами

```python
import random 

b = random.randint(1, 100)

if 71 < b <= 100:
    print(f"Ви отримали оцінку A, ваш бал: {b}")
elif 61 < b <= 70:
    print(f"Ви отримали оцінку B, ваш бал: {b}")
elif 51 < b <= 60:
    print(f"Ви отримали оцінку C, ваш бал: {b}")
elif 41 < b <= 50:
    print(f"Ви отримали оцінку D, ваш бал: {b}")
else:
    print(f"Ви отримали оцінку F, ваш бал: {b}")
```

```python
def evaluate_grade(b: int) -> str:
    if 71 < b <= 100:
        return f"Ви отримали оцінку A, ваш бал: {b}"
    elif 61 < b <= 70:
        return f"Ви отримали оцінку B, ваш бал: {b}"
    elif 51 < b <= 60:
        return f"Ви отримали оцінку C, ваш бал: {b}"
    elif 41 < b <= 50:
        return f"Ви отримали оцінку D, ваш бал: {b}"
    else:
        return f"Ви отримали оцінку F, ваш бал: {b}"
```

```python
group_of_studets = [(f"Імя {i}", random.randint(1, 100)) for i in range(1, 11)]
for b in group_of_studets:
    print(f"Студент: {b[0]}, Бал: {b[1]}, {evaluate_grade(b[1])}")
```
