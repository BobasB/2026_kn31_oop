- все в Python є об'єктом, навіть числа і стрічки, і до них можна застосувати операцію . (крапка) і викликати методи, які є у цих об'єктів
- наступні дві операції є еквівалентними, але друга є обєктно-орієнтованим підходом, а перша - процедурним підходом

```python
print(m1 + 5) # виклик оператора + для об'єкта m1, який є числом
print(m1.__add__(5)) # виклик методу __add__() для об'єкта m1, який є числом
```

- колекції даних - набір даних, які встановлені на конкретні місця
- індексіція - це доступ до елементів колекції за їх порядковим номером (індексом)
- зворотня індексація - це доступ до елементів колекції з кінця, де останній елемент має індекс -1, передостанній -2 і так далі

```python
s1 = "Це є колекція"
print(s1[0], s1[1], s1[2], s1[3], s1[4], s1[5], s1[6], s1[7], s1[8], s1[9], s1[10], s1[11], s1[12])
print(s1[0] + s1[1] + s1[2] + s1[3] + s1[4] + s1[5] + s1[6] + s1[7] + s1[8] + s1[9] + s1[10] + s1[11] + s1[12])
print(s1[-1], s1[-2], s1[-3], s1[-4], s1[-5], s1[-6], s1[-7], s1[-8], s1[-9], s1[-10], s1[-11], s1[-12], s1[-13])

print("Останній елемент: " , s1[12])
```

- колекцї можуть змінюватись, розширюватись або зменшуватись

```python
s2 = s1 + " даних елементів"
print(s2, "Останній елемент: ", s2[-1])

# в інших мовах програмування
print("Останній елемент: ", s2[len(s2) - 1]) # в інших мовах програмування
```

- слайсинг - це доступ до підмножини елементів колекції, де ми можемо вказати початковий і кінцевий індекси, а також крок
- при інндексації перший елемент завжди включається, а останній - ні, тобто якщо ми хочемо взяти елементи з 0 по 3, то ми вказуємо 0:4

```python
print(s1[5:7])
print(s1[6:10])
print(s1[:6], s1[6:])
```

- слайси можуть бути вибірковими, тобто ми можемо вказати крок

```python
print(s2[5:15:2])
print(s2[5:16:5])
```

- питання з співбесіди - як зробити реверс колекції (паліндромом)

```python
str_for_reverse = "Це є колекція"
# класично але так не роблять
#reversed_str = str_for_reverse[0:len(str_for_reverse):-1]
reversed_str = str_for_reverse[::-1]
print(reversed_str)  # Output: яцекол є еЦ
```
